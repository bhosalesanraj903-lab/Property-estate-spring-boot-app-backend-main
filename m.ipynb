```python
finished = df['loan_status'].isin(['Fully Paid', 'Charged Off', 'Default'])
df = df[finished].copy()
df.shape
```

```python
df['target'] = df['loan_status'].isin(['Charged Off', 'Default']).astype(int)
df['target'].value_counts(normalize=True)
```

```python
missing = df.isnull().mean().sort_values(ascending=False)
missing.head(30)
```

```python
# any column more than 50% empty isn't worth keeping
too_empty = missing[missing > 0.5].index
df = df.drop(columns=too_empty)
df.shape   # watch the column count drop a lot
```

```python
df.columns.tolist()
```

```python
# things that only exist AFTER the loan is funded = leakage.
# a lender would NOT have these when deciding to approve.
leakage = [
    'total_pymnt', 'total_pymnt_inv', 'total_rec_prncp', 'total_rec_int',
    'total_rec_late_fee', 'recoveries', 'collection_recovery_fee',
    'last_pymnt_d', 'last_pymnt_amnt', 'next_pymnt_d',
    'last_credit_pull_d', 'out_prncp', 'out_prncp_inv',
    'last_fico_range_high', 'last_fico_range_low',
    'funded_amnt', 'funded_amnt_inv',          # how much got funded = after-decision
    'debt_settlement_flag', 'settlement_status', 'settlement_amount',
]

# only drop the ones that actually exist in your df (some may already be gone)
leakage = [c for c in leakage if c in df.columns]
df = df.drop(columns=leakage)
df.shape
```

```python
# a few leakage stragglers that slipped through
stragglers = ['hardship_flag', 'disbursement_method', 'debt_settlement_flag']
stragglers = [c for c in stragglers if c in df.columns]
df = df.drop(columns=stragglers)
df.shape
```

```python
# Hand-pick the features a lender would actually use at application time.
# Grouped by what they tell us about the borrower.

keep = [
    # --- the loan itself ---
    'loan_amnt',        # how much they're borrowing
    'term',             # 36 or 60 months
    'int_rate',         # rate assigned (LC's own risk pricing)
    'installment',      # monthly payment
    'grade', 'sub_grade',  # LC's letter risk grade

    # --- the borrower's income & job ---
    'annual_inc',       # stated yearly income
    'emp_length',       # years employed (stability)
    'home_ownership',   # rent / own / mortgage
    'verification_status',  # did LC verify the income?

    # --- the borrower's existing debt load ---
    'dti',              # debt-to-income ratio (key risk signal)
    'revol_bal',        # revolving balance (credit card debt)
    'revol_util',       # % of credit limit used (high = stretched)
    'open_acc',         # number of open credit lines
    'total_acc',        # total credit lines ever

    # --- credit history & red flags ---
    'fico_range_low', 'fico_range_high',  # credit score at application
    'delinq_2yrs',      # late payments in last 2 yrs
    'inq_last_6mths',   # recent credit inquiries (desperation signal)
    'pub_rec',          # public records (bankruptcies etc.)
    'pub_rec_bankruptcies',
    'mort_acc',         # mortgage accounts

    # --- context ---
    'purpose',          # why they want the loan
    'addr_state',       # geography
    'issue_d',          # date - needed for time-based split later

    # --- the answer ---
    'target',
]

keep = [c for c in keep if c in df.columns]
df = df[keep].copy()
df.shape
```

```python
# look at the columns that are text but should be numbers
df[['term', 'int_rate', 'revol_util', 'emp_length']].head(10)
```

```python
# "36 months" -> 36. only term and emp_length are text; the rate cols are already numeric.
df['term'] = df['term'].str.extract(r'(\d+)').astype(int)
df['term'].value_counts()
```

```python
# which of our chosen columns still have blanks, and how bad?
missing = df.isnull().mean().sort_values(ascending=False)
missing[missing > 0]   # only show columns that actually have missing values
```

```python
# emp_length: a blank usually means unemployed / not reported.
# filling with 0 (zero years) is a defensible interpretation.
df['emp_length'] = df['emp_length'].fillna(0)

# mort_acc: blank likely means no mortgage accounts reported -> 0 is reasonable
df['mort_acc'] = df['mort_acc'].fillna(0)

# the rest are tiny (<1%). fill with the median - safe, won't move the needle.
for col in ['revol_util', 'pub_rec_bankruptcies', 'dti', 'inq_last_6mths']:
    df[col] = df[col].fillna(df[col].median())

# confirm everything is filled now
df.isnull().mean()[df.isnull().mean() > 0]   # should print nothing
```

```python
# issue_d looks like "Dec-2015" (text). turn it into a real date.
df['issue_d'] = pd.to_datetime(df['issue_d'], format='%b-%Y', errors='coerce')
df['issue_d'].min(), df['issue_d'].max()   # see the date range we have
```

```python
# if 2016 cutoff is lopsided, push the cutoff later so train has enough history.
# 2017 is a good split point: trains on 2007-2016, tests on 2017-2018.
cutoff = '2017-01-01'
train = df[df['issue_d'] < cutoff].copy()
test  = df[df['issue_d'] >= cutoff].copy()

print("train loans:", len(train))
print("test loans :", len(test))
print("train defaults:", round(train['target'].mean(), 3))
print("test defaults :", round(test['target'].mean(), 3))
```

```python
# X = the inputs the model learns from. y = the answer (default or not).
# we drop issue_d now - it did its job (the split) and isn't a predictor.
# also drop loan_status if it's still around (it's the raw version of target).
drop_cols = ['target', 'issue_d']
drop_cols = [c for c in drop_cols if c in train.columns]

X_train = train.drop(columns=drop_cols)
y_train = train['target']

X_test = test.drop(columns=drop_cols)
y_test = test['target']

print(X_train.shape, X_test.shape)
```

```python
# which columns are text (object type)? those need converting to numbers.
X_train.select_dtypes(include='object').columns.tolist()
```

```python
# re-extract the number from emp_length, then fill blanks with 0
X_train['emp_length'] = X_train['emp_length'].astype(str).str.extract(r'(\d+)').astype(float).fillna(0)
X_test['emp_length']  = X_test['emp_length'].astype(str).str.extract(r'(\d+)').astype(float).fillna(0)
X_train['emp_length'].value_counts(dropna=False)
```

```python
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.compose import ColumnTransformer

# grade & sub_grade have a natural order -> ordinal (A<B<C means something)
ordinal_cols = ['grade', 'sub_grade']

# these have NO natural order -> one-hot (separate 0/1 column per category)
onehot_cols = ['home_ownership', 'verification_status', 'purpose', 'addr_state']

# everything else is already numeric and passes through untouched
preprocessor = ColumnTransformer(
    transformers=[
        ('ord', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1), ordinal_cols),
        ('oh', OneHotEncoder(handle_unknown='ignore'), onehot_cols),
    ],
    remainder='passthrough'   # numeric columns pass through as-is
)
```

```python
from sklearn.pipeline import Pipeline
from lightgbm import LGBMClassifier

# the model: gradient-boosted trees, handles big data + mixed features well
model = Pipeline(steps=[
    ('prep', preprocessor),
    ('clf', LGBMClassifier(
        n_estimators=300,
        learning_rate=0.05,
        num_leaves=40,
        class_weight='balanced',   # because only ~20% are defaults
        random_state=42,
        n_jobs=-1
    ))
])

# train it - this is the moment the model actually learns
model.fit(X_train, y_train)
print("done training")
```

```python
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.compose import ColumnTransformer

ordinal_cols = ['grade', 'sub_grade']
onehot_cols = ['home_ownership', 'verification_status', 'purpose', 'addr_state']

preprocessor = ColumnTransformer(
    transformers=[
        ('ord', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1), ordinal_cols),
        ('oh', OneHotEncoder(handle_unknown='ignore', sparse_output=False), onehot_cols),
    ],
    remainder='passthrough'
)
print("preprocessor ready")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import HistGradientBoostingClassifier

model = Pipeline(steps=[
    ('prep', preprocessor),
    ('clf', HistGradientBoostingClassifier(
        max_iter=300, learning_rate=0.05, max_leaf_nodes=40,
        class_weight='balanced', random_state=42
    ))
])
model.fit(X_train, y_train)
print("done training")
```

```python
from sklearn.metrics import roc_auc_score
y_pred_proba = model.predict_proba(X_test)[:, 1]
auc = roc_auc_score(y_test, y_pred_proba)
print("Test AUC:", round(auc, 4))
```

```python
from sklearn.metrics import classification_report, confusion_matrix

y_pred = (y_pred_proba >= 0.5).astype(int)
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred, digits=3))
```

```python
import pandas as pd

# Expected Loss = P(default) x exposure x loss-given-default
# LGD = fraction you actually lose when a loan defaults. ~0.65 is
# a defensible industry assumption for unsecured consumer loans.
LGD = 0.65

watchlist = X_test.copy()
watchlist['pd_score'] = y_pred_proba                       # probability of default
watchlist['actual_default'] = y_test.values                # what really happened
watchlist['expected_loss'] = y_pred_proba * watchlist['loan_amnt'] * LGD

# rank loans by how much money they're expected to cost
watchlist = watchlist.sort_values('expected_loss', ascending=False)
watchlist[['loan_amnt', 'int_rate', 'grade', 'pd_score', 'expected_loss', 'actual_default']].head(10)
```

```python
# --- Fix 1 & 2: dti has placeholder junk (-1 and 999) ---
# legitimate DTI is roughly 0 to ~60. Treat the rest as bad and cap it.
for d in [X_train, X_test]:
    # replace negatives with 0 (no debt info -> assume none)
    d['dti'] = d['dti'].clip(lower=0)
    # cap absurd highs at a sane ceiling (99th percentile is defensible)
    d['dti'] = d['dti'].clip(upper=60)

# --- Fix 3: cap extreme incomes (the $9.5M outliers) ---
# winsorize: clip income at the 99th percentile so a handful of
# fake-millionaire rows don't distort the model.
income_cap = X_train['annual_inc'].quantile(0.99)
X_train['annual_inc'] = X_train['annual_inc'].clip(upper=income_cap)
X_test['annual_inc']  = X_test['annual_inc'].clip(upper=income_cap)

print("dti range now:", X_train['dti'].min(), "to", X_train['dti'].max())
print("income cap applied at: $", round(income_cap))
print("income max now:", X_train['annual_inc'].max())
```

```python
# --- engineered features: ratios that encode real credit intuition ---

# 1. installment-to-income: what share of yearly income goes to payments?
#    high = borrower stretched thin = higher risk
X_train['inst_to_income'] = (X_train['installment'] * 12) / (X_train['annual_inc'] + 1)
X_test['inst_to_income']  = (X_test['installment'] * 12) / (X_test['annual_inc'] + 1)

# 2. fico_avg: midpoint of the FICO range as one clean score
X_train['fico_avg'] = (X_train['fico_range_low'] + X_train['fico_range_high']) / 2
X_test['fico_avg']  = (X_test['fico_range_low'] + X_test['fico_range_high']) / 2

# 3. open-to-total accounts: credit history depth signal
X_train['open_to_total_acc'] = X_train['open_acc'] / (X_train['total_acc'] + 1)
X_test['open_to_total_acc']  = X_test['open_acc'] / (X_test['total_acc'] + 1)

# confirm they exist now
for f in ['inst_to_income', 'fico_avg', 'open_to_total_acc']:
    print(f, "->", f in X_train.columns)
print("X_train shape:", X_train.shape)
```

```python
from sklearn.calibration import CalibratedClassifierCV
from sklearn.metrics import roc_auc_score

# CalibratedClassifierCV wraps your pipeline and learns to correct the
# inflated probabilities. cv=3 means it uses 3-fold internal splitting
# on the training data to fit the calibration curve.
# isotonic = flexible, non-parametric calibration (good when you have lots of data)
calibrated_model = CalibratedClassifierCV(model, method='isotonic', cv=3)

print("calibrating... (this refits on training data, ~1-2 min)")
calibrated_model.fit(X_train, y_train)
print("done")

# get the NEW, calibrated probabilities on the test set
pd_calibrated = calibrated_model.predict_proba(X_test)[:, 1]

# AUC should stay about the same (~0.72) - calibration fixes probabilities,
# not ranking
print("AUC (should still be ~0.72):", round(roc_auc_score(y_test, pd_calibrated), 4))
```
